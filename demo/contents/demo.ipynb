# jupyter-webmcp

This JupyterLite site runs the [jupyter-webmcp](https://github.com/jtpio/jupyter-webmcp) extension. The extension registers [WebMCP](https://webmachinelearning.github.io/webmcp/) tools that browser agents can call to work with the notebooks, files and kernels of this page. Everything runs in the browser.

## Enable WebMCP

1. Use Chrome 155 or later.
2. Turn on `chrome://flags/#enable-webmcp-testing` and restart Chrome.
3. Reload this page. The status bar shows `WebMCP: 21 tools`: click it to list the tools. In the Notebook interface, run **Show WebMCP Tools** from the command palette.

## Connect an agent

Give your agent a browser bridge with WebMCP support, such as [Playwright MCP](https://github.com/jtpio/jupyter-webmcp#playwright-mcp) or [chrome-devtools-mcp](https://github.com/jtpio/jupyter-webmcp#chrome-devtools-mcp). Then ask, for example:

> Open https://jtpio.github.io/jupyter-webmcp/lab/index.html?path=demo.ipynb and use the WebMCP tools of the page: run the cells of demo.ipynb, then add a cell that plots the monthly revenue of each region.

JupyterLite keeps the files in the storage of the browser: when the bridge starts its own browser, the agent does not see the changes of this tab.
